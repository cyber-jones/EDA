**Missing values summary**

```python
missing_count = df.isna().sum().sort_values(ascending=False)
missing_percent = (df.isna().mean() * 100).sort_values(ascending=False)

missing_summary = pd.DataFrame({"missing_count": missing_count, "missing_percent": missing_percent})

print("Missing values summary")
missing_summary
```

**Duplicates**

```python
duplicaties_mask = df.duplicated()
num_duplicates = duplicaties_mask.sum()

print("Number of duplicate rows:", num_duplicates)

# if you want to remove duplicates:
# df_no_duplicates = df.drop_duplicates()
```

```python
expected_types = {
    "Survived": "int64",
    "Pclass": "int64",
    "Sex": "category",
    "Age": "float64",
    "Fare": "float64",
    "Embarked_Town": "category",
    "PassengerId": "int64",
}

print("Data types validation")
for col, expected in expected_types.items():
    actual = df[col].dtype
    if actual != expected:
        print(f"Column '{col}' has type '{actual}', expected '{expected}'")
    else:
        print(f"Column '{col}' has the expected type '{expected}'")
```

**Const and Quasi-Comstant columns**

```python
n_rows = len(df)
nunique = df.nunique()

constant_cols = nunique[nunique == 1].index.to_list()
print("Constant columns: ", constant_cols)

# quasi const columns
quasi_const_cols = []

for col in df.columns:
    top_freq = df[col].value_counts(normalize=True, dropna=False).values[0]
    if top_freq > 0.95 and col not in constant_cols:
        quasi_const_cols.append(col)
        
print("Quasi constant columns are (top values more than 95 percent: )", quasi_const_cols)
```

**Id like columns**

```python
n_rows = len(df)

id_like_cols = []

for col in df.columns:
    if df[col].nunique(dropna=False) == n_rows:
        id_like_cols.append(col)
        
print("Id like columns:", id_like_cols)
```

**String inconsistency**

```python
object_cols = df.select_dtypes(include=["object"]).columns.to_list()
print("Object or category based columns", object_cols)

# Simple clean: strip spaces convert to lower case
df["Embarked_Town_Clean"] = (
    df["Embarked_Town_Dirty"]
    .astype(str)
    .str.strip()
    .str.lower()
    .replace("unknown", np.nan)
)
```

**High null columns**

```python
high_null_threashold = 0.4 # 40% are missing in a column

high_null_cols = missing_summary[missing_summary["missing_percent"] >= high_null_threashold * 100]

print("Columns with high missing percentage:")
print(high_null_cols)
```

**High zero columns (numeric)**

```python
numeric_cols = df.select_dtypes(include=[np.number]).columns.to_list()
zero_share = {}

for col in numeric_cols:
    zero_share[col] = (df[col] == 0).mean()

zero_share_series = pd.Series(zero_share).sort_values(ascending=False)
print(zero_share)

print("-"*50)

high_zero_threashold = 0.8
high_zero_cols = zero_share_series[zero_share_series >= high_zero_threashold]
print("Numeric columns with many zeros:", high_zero_cols)
```
